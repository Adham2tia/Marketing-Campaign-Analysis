Function to clean Categorical Columns

```python
cat_cols = ['FocusCategory','TargetAudience','PromoCode']
def clean_cat_cols(df,columns):
    for col in columns:
        df[col] = ((
            df)[col]
            .astype(str)
            .str.strip()
                            )
    return df
cleaned_columns = clean_cat_cols(DimCampaign,cat_cols)
cleaned_columns
```

```python
DimCampaign[cat_cols] = DimCampaign[cat_cols].replace({
    'NaN': 'unknown',
    '': 'unknown',
    'None': 'unknown',
})
DimCampaign.head(10)
```

```python
DimCampaign['PromoCode'] = DimCampaign['PromoCode'].fillna('No Promo')
DimCampaign.head(10)
```

Function Detect_Outlier

```python
def detect_outliers(df, col, return_count=False):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    if return_count:
        return len(outliers)
    return outliers
```

```python
outliers = detect_outliers(DimCampaign, 'BudgetEGP')
outliers
```
